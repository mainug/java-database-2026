#### 일반적인 방법

- 접속 후
- 비지니스 트랜잭션 처리가 끝나면
- 접속 종료

````python
import oracledb

dsn = 'localhost:1521/XE'

conn = oracledb.connect(
    user = 'test',
    password = 'test12345',
    dsn = dsn
)

print('Oracle 연결 성공!', conn.version)

# 커서 생성
cursor = conn.cursor()

# SQL문 실행, SQL문에 세미콜론(;) 넣지 말 것
sql = """
SELECT DEPT_ID, DEPT_NAME, LOC
  FROM DEPT
 ORDER BY DEPT_ID
"""

# 실행
cursor.execute(sql)

# 결과 출력
for row in cursor:
    print(row)

cursor.close()
conn.close()

print("Oracle 접속 종료.")
````

#### INSERT 실행

- DBeaver에서 TEST 스키마에 SEQ_DEPT 시퀀스 생성

````python
import oracledb

dsn = 'localhost:1521/XE'

conn = oracledb.connect(
    user = 'test',
    password = 'test12345',
    dsn = dsn
)

print('Oracle 연결 성공!', conn.version)

try:
    # 커서 생성
    cursor = conn.cursor()

    # INSERT 쿼리 작성
    sql = """
    INSERT INTO DEPT (DEPT_ID, DEPT_NAME, LOC)
    VALUES (SEQ_DEPT.NEXTVAL, :1, :2)
    """

    # INSERT 쿼리 실행
    dept_name = '생산부'; loc = 'KWANGJU'
    cursor.execute(sql, (dept_name, loc))

    #트랜잭션 처리
    conn.commit()
except Exception as e:
    print('예외 발생 ', e.args)
    conn.rollback()
finally:
    cursor.close()
    conn.close()

print('INSERT 완료')
````

#### WHERE 조건 실행

````python
import oracledb

dsn = 'localhost:1521/XE'

conn = oracledb.connect(
    user = 'test',
    password = 'test12345',
    dsn = dsn
)

print('Oracle 연결 성공!', conn.version)

# 커서 생성
cursor = conn.cursor()

# SQL문 실행, SQL문에 세미콜론(;) 넣지 말 것
sql = """
SELECT DEPT_ID, DEPT_NAME, LOC
  FROM DEPT
 WHERE LOC = :LOC
 ORDER BY DEPT_ID
"""

# 실행, 조건절 파라미터 추가
cursor.execute(sql, LOC='SEOUL')
rows = cursor.fetchall() # 데이터의 양이 작을 때 한 번에 모두 가져오기

# 결과 출력
for row in rows:
    print(row)

# 종료
cursor.close()
conn.close()

print("Oracle 접속 종료.")
````

#### UPDATE 실행

````python
import oracledb

dsn = 'localhost:1521/XE'

conn = oracledb.connect(
    user = 'test',
    password = 'test12345',
    dsn = dsn
)

print('Oracle 연결 성공!', conn.version)

try:
    # 커서 생성
    cursor = conn.cursor()

    # UPDATE 쿼리 작성
    sql = """
    UPDATE DEPT
       SET DEPT_NAME = :1
         , LOC = :2
     WHERE DEPT_ID = :3
    """

    # UPDATE 쿼리 실행
    DEPT_NAME = '자재부'
    LOC = 'INCHEON'
    DEPT_ID = 50
    cursor.execute(sql, (DEPT_NAME, LOC, DEPT_ID))

    #트랜잭션 처리
    conn.commit()
except Exception as e:
    print('예외 발생 ', e.args)
    conn.rollback()
finally:
    cursor.close()
    conn.close()

print('UPDATE 완료')
````

#### DELETE 실행

````python
import oracledb

dsn = 'localhost:1521/XE'

conn = oracledb.connect(
    user = 'test',
    password = 'test12345',
    dsn = dsn
)

print('Oracle 연결 성공!', conn.version)

try:
    # 커서 생성
    cursor = conn.cursor()

    # DELETE 쿼리 작성
    sql = """
    DELETE FROM DEPT
     WHERE DEPT_ID = :1
    """

    # UPDATE 쿼리 실행
    DEPT_ID = 50
    cursor.execute(sql, (DEPT_ID,))
    print('삭제 행 : ', cursor.rowcount)

    #트랜잭션 처리
    conn.commit()
except Exception as e:
    print('예외 발생 ', e.args)
    conn.rollback()
finally:
    cursor.close()
    conn.close()

print('DELETE 완료')
````
